# Synthethic Generation

```python
import h5py
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

## 1. Elipse Generation with random texture

In this section the tumor will be elipses with a random size and a random texture

```python
def elipse_generation(img_path):
    # read image h5 
    with h5py.File(img_path, 'r') as f:
        img = f['data'][()]

    # get image size
    image_size = img.shape[0]

    # Ellipse parameters
    center = (np.random.randint(image_size/3, 2*image_size/3), np.random.randint(image_size/3, 2*image_size/3))  # Random center (x, y)
    axes = (np.random.randint(5, 10), np.random.randint(5, 10))  # Random axes (major axis, minor axis)
    angle = np.random.randint(0, 180)  # Random rotation angle in degrees
    start_angle = 0
    end_angle = 360  # Full ellipse

    color = np.random.randint(6, 10)  # White color

    # Draw the ellipse
    cv2.ellipse(img, center, axes, angle, start_angle, end_angle, color, -1)  # -1 fills the ellipse

    mask = np.zeros((image_size, image_size), dtype=np.uint8)
    cv2.ellipse(mask, center, axes, angle, start_angle, end_angle, 255, -1)  # -1 fills the ellipse

    # Show the image
    plt.imshow(img, cmap='gray')
    plt.axis("off")
    plt.show()

    return img, mask
```
