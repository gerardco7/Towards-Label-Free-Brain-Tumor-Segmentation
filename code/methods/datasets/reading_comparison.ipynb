```python
import os
import h5py
import time
import nibabel as nib


from DatasetBlosc2 import DatasetBlosc2
```

```python
# Obtain 100 train slices

# path to the train data
old_path = 'D:/data/BraTS2023-MEN-adapted/train/'
new_path = 'D:/data/BraTS2023-MEN-adapted-Blosc2-zscore_brain/train/'
original_path = 'D:/data/BraTS2023-MEN'

# the old path contains slices, the new path contains volumes
```

```python
# Old path
start = time.time() 

for i in range(100):
    index = 0
    for modality in os.listdir(old_path):
        modality_path = os.path.join(old_path, modality).replace('\\', '/')
        for slices in os.listdir(modality_path):
            slice_path = os.path.join(modality_path, slices).replace('\\', '/')
            # read slice h5py
            with h5py.File(slice_path, 'r') as f:
                slice = f['data'][:]

            index += 1

            if index == 100:
                break
        if index == 100:
            break

print('Old path time:', (time.time() - start)*4)
```

```python
# New path
start = time.time()
data = DatasetBlosc2(new_path)

for i in range(100):
    index = 0 

    for volume in os.listdir(new_path):
        identifier = volume.split('.')[0]
        read = data.load_case(identifier=identifier)
        assert read[0] == read[1]
        for i in range(read[0].shape[-1]):
            index += 1
            if index >= 100:
                break
            
        if index >= 100:
            break

print('New path time:', time.time() - start)
```

```python
# Llegir dades originals

start = time.time()

for i in range(100):
    index = 0

    for patient in os.listdir(original_path):
        patient_path = os.path.join(original_path, patient).replace('\\', '/')
        for modality in os.listdir(patient_path):
            if 'seg' not in modality:
                modality_path = os.path.join(patient_path, modality).replace('\\', '/')
                # read volume with nibabel
                volume = nib.load(modality_path).get_fdata()

        break

print('Original path time:', time.time() - start)
```
