```python
import sys
import os
import matplotlib.pyplot as plt
import numpy as np

sys.path.append(os.path.abspath('../code/methods/datasets'))
sys.path.append(os.path.abspath('../code/methods/losses'))
sys.path.append(os.path.abspath('../code'))

from DatasetBlosc2 import DatasetBlosc2
```

```python
dataset_path = 'D:/data/BraTS2023-MEN-adapted-Blosc2-zscore_brain/test'

dataset = DatasetBlosc2(folder=dataset_path,
                        identifiers=None)
```

```python
idx = dataset.identifiers
data, seg = dataset.load_case(idx[6])

# plot 
fig, ax = plt.subplots(1, 5, figsize=(15, 5))
ax[0].imshow(seg[:, :, 22], cmap='gray')
ax[0].set_title('seg')
ax[1].imshow(data[0, :, :, 22], cmap='gray')
ax[1].set_title('t1c')
ax[2].imshow(data[1, :, :, 22], cmap='gray')
ax[2].set_title('t1n')
ax[3].imshow(data[2, :, :, 22], cmap='gray')
ax[3].set_title('t2f')
ax[4].imshow(data[3, :, :, 22], cmap='gray')
ax[4].set_title('t2w')
plt.show()
```

```python
t1c = data[0]
t1n = data[1]
t2f = data[2]
t2w = data[3]
```

```python
t2f_t1c = t2f - t1c

# Plot t1c, t2f, and t1c_t2f as heatmaps
fig, ax = plt.subplots(1, 3, figsize=(15, 5))
cmap = 'hot'  # Define the colormap

im0 = ax[0].imshow(t1c[:, :, 20], cmap=cmap)
ax[0].set_title('t1c')
fig.colorbar(im0, ax=ax[0])

im1 = ax[1].imshow(t2f[:, :, 20], cmap=cmap)
ax[1].set_title('t2f')
fig.colorbar(im1, ax=ax[1])

im2 = ax[2].imshow(t2f_t1c[:, :, 20], cmap=cmap)
ax[2].set_title('t1c - t2f')
fig.colorbar(im2, ax=ax[2])

plt.show()
```

```python
# sum t1c and t2f
t1c_t2f_sum = t1c + t2f

# Plot t1c, t2f, and t1c_t2f_sum as heatmaps
fig, ax = plt.subplots(1, 3, figsize=(15, 5))
cmap = 'hot'  # Define the colormap

im0 = ax[0].imshow(t1c[:, :, 20], cmap=cmap)
ax[0].set_title('t1c')
fig.colorbar(im0, ax=ax[0])

im1 = ax[1].imshow(t2f[:, :, 20], cmap=cmap)
ax[1].set_title('t2f')
fig.colorbar(im1, ax=ax[1])

im2 = ax[2].imshow(t1c_t2f_sum[:, :, 20], cmap=cmap)
ax[2].set_title('t1c + t2f')
fig.colorbar(im2, ax=ax[2])
plt.show()

```
