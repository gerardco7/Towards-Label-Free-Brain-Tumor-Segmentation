```python
import sys
import os
import matplotlib.pyplot as plt
import numpy as np

sys.path.append(os.path.abspath('../code/methods/datasets'))
sys.path.append(os.path.abspath('../code/methods/losses'))
sys.path.append(os.path.abspath('../code'))

from DatasetBlosc2 import DatasetBlosc2
from losses import ssim_loss
```

```python
dataset_path = 'D:/data/BraTS2023-MEN-adapted-Blosc2/train'

dataset = DatasetBlosc2(folder=dataset_path,
                        identifiers=None)
```

```python
idx = dataset.identifiers
data, seg = dataset.load_case(idx[0])
```

```python
# plot 
fig, ax = plt.subplots(1, 4, figsize=(15, 5))
ax[0].imshow(data[0, :, :, 70], cmap='gray')
ax[0].set_title('t1c')
ax[1].imshow(data[1, :, :, 70], cmap='gray')
ax[1].set_title('t1n')
ax[2].imshow(data[2, :, :, 70], cmap='gray')
ax[2].set_title('t2w')
ax[3].imshow(data[3, :, :, 70], cmap='gray')
ax[3].set_title('t2f')
plt.show()
```

```python
# Generate a random tensor
import torch
import torch.nn.functional as F

data = data.clone().detach().to(dtype=torch.float32)
r = torch.rand_like(data)

ssim = ssim_loss(data[0].unsqueeze(0), r[0].unsqueeze(0))
```

```python
ssim
```

```python
# plot the random tensor and the ssim
fig, ax = plt.subplots(1, 2, figsize=(10, 5))
ax[0].imshow(r[0, :, :, 70], cmap='gray')
ax[0].set_title('Random Tensor')
ax[1].imshow(data[0, :, :, 70], cmap='gray')
ax[1].set_title('data')
plt.show()
```

```python
# take the minimum of data[0] and create an image with the same size as data[0]
min_value = torch.min(data[0])
image_with_min = torch.full_like(data[0], min_value)


# compute ssim 
ssim = ssim_loss(data[0].unsqueeze(0), image_with_min.unsqueeze(0))
ssim
```

```python
# plot the image with the minimum value
fig, ax = plt.subplots(1, 2, figsize=(10, 5))
ax[0].imshow(image_with_min[:, :, 20], cmap='gray')
ax[0].set_title('Image with Minimum Value')
ax[1].imshow(data[0, :, :, 20], cmap='gray')
ax[1].set_title('data')
plt.show()
```
